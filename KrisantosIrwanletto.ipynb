```python

# PROGRAM PENYELESAIAN SPL 3 VARIABEL
# METODE ELIMINASI GAUSS

# 1. Memasukan matriks persamaan
A = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

print("MATRIKS AWAL:")
for baris in A:
    print(baris)


# 2. Proses Eliminasi Gauss
n = len(A)

for i in range(n):


    # Mencari nilai pivot terbesar
    pivot = max(range(i, n), key=lambda k: abs(A[k][i]))

    # Menukar baris
    A[i], A[pivot] = A[pivot], A[i]

    if abs(A[i][i]) < 1e-12:
        raise ValueError("Sistem tidak memiliki solusi tunggal")


    # Menghilangkan nilai di bawah pivot
    for j in range(i + 1, n):
        faktor = A[j][i] / A[i][i]

        for k in range(i, n + 1):
            A[j][k] = A[j][k] - faktor * A[i][k]


# 3. Subtitusi Mundur
x = [0] * n

for i in range(n - 1, -1, -1):
    jumlah = 0

    for j in range(i + 1, n):
        jumlah += A[i][j] * x[j]

    x[i] = (A[i][n] - jumlah) / A[i][i]


# 4. Menampilkan Hasil
print("\nHASIL PENYELESAIAN SPL:")
print("Nilai x =", x[0])
print("Nilai y =", x[1])
print("Nilai z =", x[2])
```
